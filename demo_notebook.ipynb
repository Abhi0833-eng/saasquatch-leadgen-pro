4. **Export to CSV & CRM JSON**

```python
import json
import pandas as pd

class CapraeSourcingEngine:
    def calculate_deal_score(self, lead):
        score = 0
        factors = []
        
        # EBITDA Range ($1M - $5M is optimal for Search Funds / ETA)
        ebitda = lead.get('ebitdaNumeric', 0)
        if 1000000 <= ebitda <= 5000000:
            score += 25
            factors.append('Optimal ETA EBITDA Range ($1M-$5M)')
        elif ebitda > 5000000:
            score += 15
            
        # Founder Retirement Signal (Age 58+)
        age = lead.get('founderAge', 0)
        if age >= 60:
            score += 30
            factors.append(f'High Founder Exit Motivation (Age: {age})')
        elif age >= 55:
            score += 20
            
        # Caprae AI Value Creation Upside (Low digital score = massive post-acq upside)
        digital = lead.get('digitalMaturityScore', 50)
        if digital < 45:
            score += 30
            factors.append(f'Prime Caprae AI Transformation Opportunity (Digital: {digital}/100)')
        elif digital < 65:
            score += 15
            
        # Operating Margin
        if float(lead.get('ebitdaMargin', '0%').replace('%','')) >= 25:
            score += 15
            factors.append('Strong Cash-Flow Operating Margin (>=25%)')
            
        lead['dealScore'] = min(score, 99)
        lead['dealRationale'] = factors
        return lead

engine = CapraeSourcingEngine()
print('Engine Initialized Successfully.')
```
